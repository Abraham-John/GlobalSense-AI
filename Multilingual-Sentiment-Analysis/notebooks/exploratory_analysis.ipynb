# GlobalSense AI - Multilingual Sentiment Corpus Exploratory Analysis

This notebook documents the empirical properties of the 10-language balanced sentiment corpus.
Languages: English, Hindi, Hinglish, French, Spanish, German, Italian, Portuguese, Arabic, Japanese.
Classes: Positive, Negative, Neutral (3-class classification).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json

# Load processed datasets
train_df = pd.read_csv('../backend/dataset/processed/train.csv')
val_df = pd.read_csv('../backend/dataset/processed/val.csv')
test_df = pd.read_csv('../backend/dataset/processed/test.csv')

print(f'Train samples: {len(train_df):,}')
print(f'Validation samples: {len(val_df):,}')
print(f'Test samples: {len(test_df):,}')
```

## 1. Class Distribution Across Splits

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, df, split in zip(axes, [train_df, val_df, test_df], ['Train', 'Validation', 'Test']):
    counts = df['label'].value_counts()
    sns.barplot(x=counts.index, y=counts.values, ax=ax, palette='mako')
    ax.set_title(f'{split} Set Class Balance')
    ax.set_ylabel('Number of Samples')
plt.tight_layout()
plt.show()
```

## 2. Language Representation & Minimum Threshold (>= 200) Verification

```python
test_lang_counts = test_df['language'].value_counts()
plt.figure(figsize=(10, 4))
bars = plt.bar(test_lang_counts.index, test_lang_counts.values, color='#4f46e5')
plt.axhline(200, color='red', linestyle='--', label='Min Supported Threshold (>=200)')
plt.title('Held-Out Test Samples per Language')
plt.xlabel('Language Code')
plt.ylabel('Test Samples')
plt.legend()
plt.show()

print('Supported Audit:')
for lang, cnt in test_lang_counts.items():
    status = 'SUPPORTED' if cnt >= 200 else 'LIMITED'
    print(f'  {lang}: {cnt} samples -> {status}')
```

## 3. Sequence Length Distribution by Language

```python
train_df['word_count'] = train_df['text'].apply(lambda x: len(str(x).split()))
plt.figure(figsize=(12, 5))
sns.boxplot(data=train_df, x='language', y='word_count', palette='viridis')
plt.title('Word Count Distribution Across Target Languages')
plt.ylim(0, 150)
plt.show()
```
